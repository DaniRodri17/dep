```python
import pandas as pd
import numpy as np
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Dropout
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, mean_squared_error, mean_absolute_error
from sklearn.ensemble import RandomForestClassifier
```

```python
df = pd.read_csv('penguins_size.csv')
df.head()
```

```python
df.dropna(inplace=True)
```

```python
X = df.drop('species', axis=1)
y = df['species']

# Apply one-hot encoding to categorical columns 'island' and 'sex'
X = pd.get_dummies(X, columns=['island', 'sex'], drop_first=True)

# Convert target variable 'species' to one-hot encoded format
y = pd.get_dummies(y)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
dt = RandomForestClassifier()
dt.fit(X_train, y_train)
dt_pred = dt.predict(X_test)
print(accuracy_score(y_test, dt_pred))
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, dt_pred))
```

```python
from sklearn.metrics import confusion_matrix
print(confusion_matrix(y_test, dt_pred))
#
```

```python
model = Sequential([
    Dense(512, activation='relu', input_shape=(8,)),
    Dense(312, activation='relu'),

    Dense(3, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['precision']
)
```

```python
history = model.fit(x=X_train, y=y_train,
                    validation_data=(X_test, y_test),
                    batch_size=10,
                    epochs=200)
```
